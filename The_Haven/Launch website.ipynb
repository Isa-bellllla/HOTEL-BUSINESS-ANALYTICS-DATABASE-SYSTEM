```python
from pathlib import Path
import subprocess
import sys
from IPython.display import Markdown, display

project = Path.home() / "Desktop" / "The_Haven_Los_Angeles"
launcher = project / "launch_hotel.py"

if not launcher.is_file():
    raise FileNotFoundError(f"Update the project path. Could not find {launcher}")

hotel = subprocess.Popen(
    [sys.executable, "-I", "-B", str(launcher), "--no-browser"],
    cwd=project,
    stdout=subprocess.PIPE,
    text=True,
)

for line in hotel.stdout:
    if "http://127.0.0.1:" in line:
        url = line.strip().split()[-1]
        display(Markdown(f"[Open The Haven]({url})"))
        break
else:
    raise RuntimeError("The hotel server exited before it could start.")
```
